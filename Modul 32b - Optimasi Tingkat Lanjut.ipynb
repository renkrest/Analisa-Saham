## 1. Genetic Algorithm

*Genetic Algorithm* atau disingkat GA merupakan sebuah algoritma yang terinspirasi dari proses mutasi genetik dan merupakan salah satu bagian dari keilmuan *artificial intelligence*.<br>
GA dapat digunakan untuk menyelesaikan banyak tipe permasalahan.

Di konteks lain, GA sering juga dikenal sebagai *evolutionary algorithm* dan merupakan pendekatan *metaheuristic*.



<br>

### 1.1 Penentuan solusi awal

Konsep sederhananya dapat dilihat melalui ilustrasi berikut:

<img src="https://i.ibb.co/bRj4NDH/ga.jpg" alt="ga" border="0">

Di sebelah kiri ada S1 di mana S1 adalah ilustrasi untuk solusi ke-1, di mana pada tahapan awal ini solusinya ini didapatkan secara *random*.<br>
Di GA, kita bisa mengembangkan solusi awal, dan tentunya jumlah solusinya bisa banyak sampai solusi ke-n (Sn).<br>
Kemudian dari setiap solusi tersebut akan masuk ke tahapan selanjutnya yaitu *fitness function*, di mana pada bagian ini akan dilihat apakah setiap solusi awal yang dihasilkan tersebut sudah cukup baik atau tidak.<br>
Dalam konteks ini *fitness function* mirip dengan *objective function* yang sebelumnya dipelajari di modul sebelum ini.

Jika sudah, maka selanjutnya akan mengalami proses *evolutionary approach*, di mana solusi tadi mengalami evolusi (layaknya evolusi genetik pada mahluk hidup).<br>
Selanjutnya proses itu akan terus berulang melalui banyak iterasi sampai didapatkan solusi layak terbaik.

Setiap solusi *random* (misal S1) disebut sebagai *initial population* / *initial solution*.

Perlu diperhatikan juga bahwa GA bukanlah model *fixed* layaknya *linear programming*.<br>
GA lebih layak disebut sebagai *process algorithm*, di mana kita melakukan iterasi berkali-kali sampai didapatkan solusi layak optimumnya.

### 1.2 *Fitness evaluation*

Sekarang anggap setelah melalui tahapan *fitness evaluation* (evaluasi dari solusi yang ada), kita mendapatkan hasil sebagai berikut:
- individu 1 = 17
- individu 2 = 15
- individu 3 = 11

Pada dasarnya untuk mendapatkan nilai evaluasi dari setiap individu tersebut, maka kita cukup memasukkan nilai dari setiap variabel ke *objective function* (fungsi tujuan).<br>
Sebagai contoh misal fungsi tujuan kita adalah X+Y, maka cukup memasukkan nilai X dan nilai Y kita akan mendapatkan nilai fungsi tujuannya.

<br>

### 1.3 *Evolutionary approach*

Pada tahapan *evolutionary approach*, ada 2 proses yang terjadi yaitu:
1. pencampuran (*crossover*)
2. mutasi (*mutation*)

<img src="https://i.ibb.co/fH22xTV/ga3.jpg" alt="ga3" border="0">

Di tahapan sebelumnya kita mengetahui bahwa solusi 1 (individu 1) merupakan solusi terbaik karena nilai *objective*-nya paling tinggi.<br>

Kemudian di bagian pencampuran (*crossover*), maka kita memilih 2 individu (disebut sebagai *parents*) di mana salah satunya adalah individu terbaik tadi, dan satunya adalah individu dengan performa yang cukup baik.<br>
Kemudian kita mencampur keduanya menjadi solusi baru (disebut sebagai *children*), di mana dalam konsep biologi maka ini adlaah proses pembuahan (reproduksi).

Sekarang pada proses mutasi, maka salah satu gen dari *children* ini bermutasi (berubah) menjadi sedikit berbeda, di mana perubahannya terjadi secara *random*.

<br>

### 1.4 *Stopping criteria*

Proses pencampuran dan mutasi dilanjutkan dengan mencampurkan lagi dengan *parents* sebelumnya sehingga membentuk jumlah solusi yang sama (pembentukan generasi kedua).

Kemudian dilanjutkan dengan proses awal, mulai dari penentuan solusi awal, pencampuran, mutasi, penyusunan solusi gabungan, dan seterusnya sampai mencapai kriteria tertentu.

Penentuan kriteria berhentinya proses iterasi GA antara lain:
- jumlah generasi terbentuk tercapai sampai jumlah tertentu
- jumlah generasi terbentuk solusinya tidak lagi bertambah baik
- kualitas dari solusi yang didapatkan

Sekarang kita akan coba eksekusi *genetic algorithm* di python.

Untuk dapat menjalankan GA di python pastikan sudah menginstall *library*-nya dengan cara sebagai berikut:<br>

```python
# Mengimpor library
import numpy as np
from geneticalgorithm import geneticalgorithm as ga
```

Sekarang kita definisikan variabel yang ingin kita cari solusinya-nya:

```python
# Mendefinisikan variabel
varbounds = np.array([[0,10],[0,10]])
vartype = np.array([['int'],['real']])
```

Pada perintah `varbounds` kita sedang mendefinisikan 2 variabel (ada dua tanda kurung di dalamnya), di mana setiap variabel memiliki rentang solusi antara 0 hingga 10.<br>
Kemudian pada `vartype` kita mendefinisikan tipe datanya. Untuk variabel pertama bertipe integer (tanpa desimal), sementara yang kedua adalah real (bisa desimal).

Variabel keputusan yang kita definisikan akan menggunakan simbol `x`, maka kita memiliki `x[0]` dan `x[1]`.

Sekarang kita akan mendefinisikan fungsi dari GA:

```python
# Mendefinisikan fungsi GA
def f(x):

    # Sekarang kita definisikan fungsi pembatas
    # pen adalah penalization (pemberian penalty) yang secara default nilainya 0
    # Jika ada fungsi pembatas yang tidak tercapai, maka akan kena penalti
    pen = 0
    if not -x[0]+2*x[1]*x[0]<=8: 
        pen = np.inf
    if not 2*x[0]+x[1]<=14: 
        pen = np.inf
    if not 2*x[0]-x[1]<=10: 
        pen = np.inf
    
    # Secara default fungsi tujuan GA adalah minimasi,
    # Maka jika ingin maksimasi tambahkan tanda minus di bagian return
    return -(x[0]+x[1]*x[0]) + pen
```

Sekarang kita definisikan parameter algoritmanya, termasuk di dalamnya adalah *stopping criteria*.

```python
# Mendefinisikan parameter algoritmanya
algorithm_param = {'max_num_iteration': 100,
                   'population_size':100,
                   'mutation_probability':0.1,
                   'elit_ratio': 0.01,
                   'crossover_probability': 0.5,
                   'parents_portion': 0.3,
                   'crossover_type':'uniform',
                   'max_iteration_without_improv':None}
```

```python
# Mendefinisikan model ga
model = ga(function=f,
           dimension=2,
           variable_type_mixed=vartype,
           variable_boundaries=varbounds,
           algorithm_parameters=algorithm_param)

# Menjalankan model ga
model.run()
```

Sekarang kita sudah mendapatkan solusi optimumnya, di mana:
- nilai `x[0]` = 5
- nilai `x[1]` = 1.28
- nilai fungsi tujuan (Z) = 11.41

Catatan penting bagi GA adalah walaupun solusi yang dihasilkan tampak optimum, namun solusi dari GA tidak menjamin bahwa ini merupakan solusi *global optimum* (bisa jadi ada solusi lain yang lebih baik, namun jaraknya kemungkinan tidak jauh dari solusi terbaik saat ini).

## 2. *Particle Swarm Optimization* (PSO)

PSO merupakan algoritma yang mirip dengan GA, di mana ia juga masuk ke dalam algoritma metaheuristik.

Ilustrasi mudahnya bisa dilihat melalui gambar berikut:

<img src="https://i.ibb.co/zGDy3G1/pso.jpg" alt="pso" border="0">

PS (*particle swarm*) artinya adalah kumpulan / kawanan. Ilustrasi sederhananya sesuai gambar adalah kawanan lebah yang terbang mencari bunga.<br>
Dari kumpulan ini, beberapa lebah akan menyebar mencari bunga.

Pada gambar di atas terdapat garis putus-putus dan garis tebal.<br>
Garis putus-putus adalah lintasan lebah, sementara garis tebal adalah arah lebah dan kecepatan lebah selanjutnya untuk mencari bunga.

Anggap terdapat 3 lebah, di mana:
- lebah kedua sudah menemukan solusi optimal globalnya
- lebah pertama memiliki 2 arah tujuan, yang pertama adalah solusi global optimal dan satunya lagi adalah bunga pilihannya sendiri
- Lebah ketiga masih belum menemukan bunga pilihannya dan dia juga masih terbawa arah menuju global optimal

Dari ketiga lebah tersebut (yang merupakan bagian dari kawanan utuh), maka kawanan ini akan terus mengupdate kecepatan dan arah mereka dalam mencari bunga.

Di bagian akhir nanti, semua bunga sudah dihinggapi oleh kawanan lebah, dan kumpulan lebah akan berakhir pada bunga yang paling banyak dihuni kawanannya.
